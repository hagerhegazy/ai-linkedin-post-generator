```python

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
model1_name = "Qwen/Qwen3-1.7B"
model2_name = "Qwen/Qwen2.5-3B"

tokenizer1= AutoTokenizer.from_pretrained(model1_name)
tokenizer2= AutoTokenizer.from_pretrained(model2_name)
model1 = AutoModelForCausalLM.from_pretrained(
    model1_name,
    torch_dtype=torch.float16,
    device_map="auto"
)
model2 = AutoModelForCausalLM.from_pretrained(
    model2_name,
    torch_dtype=torch.float16,
    device_map="auto"
)
```

```python
def generate_response(prompt, max_new_tokens=500):

    messages = [
        {
            "role": "user",
            "content": prompt
        }
    ]

    encoded_inputs = tokenizer1.apply_chat_template(
        messages,
        tokenize=True,
        add_generation_prompt=True,
        return_tensors="pt"
    ).to(model1.device)

    outputs = model1.generate(
        encoded_inputs["input_ids"], # Pass the input_ids tensor
        max_new_tokens=max_new_tokens,
        temperature=0.7,
        do_sample=True
    )

    generated_tokens = outputs[0][encoded_inputs["input_ids"].shape[-1]:] # Use input_ids shape

    response = tokenizer1.decode(
        generated_tokens,
        skip_special_tokens=True
    )

    return response

response = generate_response(
    "Explain what an AI agent is in simple words."
)

print(response)
```

```python
def generate_review(prompt, max_new_tokens=500):

    messages = [
        {
            "role": "user",
            "content": prompt
        }
    ]

    encoded_inputs = tokenizer2.apply_chat_template(
        messages,
        tokenize=True,
        add_generation_prompt=True,
        return_tensors="pt"
    ).to(model2.device)

    outputs = model2.generate(
        encoded_inputs["input_ids"], # Pass the input_ids tensor
        max_new_tokens=max_new_tokens,
        temperature=0.7,
        do_sample=True
    )

    generated_tokens = outputs[0][encoded_inputs["input_ids"].shape[-1]:] # Use input_ids shape

    response = tokenizer2.decode(
        generated_tokens,
        skip_special_tokens=True
    )

    return response

response = generate_review(
    "Explain what an AI agent is in simple words."
)

print(response)
```

```python
writer_prompt = """
You are a professional LinkedIn content writer.

Your job is to create a LinkedIn post from the user's information.

Rules:
- Write naturally, like a real person.
- Do not sound like AI.
- Do not invent achievements or information.
- Start with an interesting hook.
- Keep paragraphs short.
- Make the post professional but engaging.
- Avoid unnecessary emojis.
- End with a meaningful takeaway or question.
- Add 3-5 relevant hashtags.
- make it small sentences not in one line

User information:

I am learning Agentic AI.
I am studying LLMs, RAG, AI agents, LangChain,
FastAPI, and related technologies.
I am working toward becoming a Full Stack AI Engineer.

Write the LinkedIn post.
"""

draft = generate_response(writer_prompt)

print(draft)
```

```python
reviewer_prompt = f"""
You are a senior LinkedIn content editor.

Review the LinkedIn post below.

Your job is to:
1. Check grammar and clarity.
2. Remove unnecessary repetition.
3. Make the hook stronger.
4. Make the writing sound natural and human.
5. Improve engagement without using clickbait.
6. Make sure the claims are realistic.
7. Keep the author's original meaning.
8. Improve the structure and readability.
9. Keep it concise.
10. Return ONLY the final improved LinkedIn post.
11. make it small sentences not in one line

DRAFT:
{draft}
"""

final_post = generate_review(reviewer_prompt)

print(final_post)
```

```python
def create_linkedin_post(user_information):

    writer_prompt = f"""
You are a professional LinkedIn content writer.

Write a LinkedIn post using ONLY the information provided.

Requirements:
- Natural and human
- Professional but not robotic
- Strong hook
- Short paragraphs
- No invented achievements
- No exaggerated claims
- Suitable for LinkedIn
- 3-5 relevant hashtags

USER INFORMATION:
{user_information}

Write the draft.
"""

    draft = generate_response(writer_prompt)

    reviewer_prompt = f"""
You are a senior LinkedIn content editor.

Review and improve the following draft.

Check:
- Grammar
- Clarity
- Hook
- Structure
- Engagement
- Natural human tone
- Repetition
- Accuracy

Do not invent information.
Preserve the author's meaning.

Return ONLY the final LinkedIn post.

DRAFT:
{draft}
"""

    final_post = generate_review(reviewer_prompt)

    return draft, final_post
```

```python
information = """
I completed an AI project where I built...
I used Python, LangChain and RAG.
The main thing I learned was...
"""

draft, final_post = create_linkedin_post(information)

print("========== DRAFT ==========")
print(draft)

print("\n========== FINAL ==========")
print(final_post)
```

## 🎨 Beautiful Colorful Gradio UI

واجهة تفاعلية ملونة تاخد معلوماتك، وتطلعلك المسودة (Draft) من الموديل الأول، والنسخة النهائية (Final) بعد المراجعة من الموديل الثاني.

```python
import gradio as gr

# ============================
# 🎨 Custom colorful theme
# ============================
custom_theme = gr.themes.Soft(
    primary_hue="violet",
    secondary_hue="pink",
    neutral_hue="slate",
    font=[gr.themes.GoogleFont("Poppins"), "sans-serif"],
).set(
    button_primary_background_fill="linear-gradient(90deg, #7F00FF, #E100FF)",
    button_primary_background_fill_hover="linear-gradient(90deg, #E100FF, #7F00FF)",
    button_primary_text_color="white",
    block_title_text_weight="700",
    block_border_width="2px",
    block_shadow="*shadow_drop_lg",
)

custom_css = """
#title-banner {
    text-align: center;
    padding: 22px 10px;
    border-radius: 18px;
    background: linear-gradient(120deg, #7F00FF, #E100FF, #FF7AC6);
    color: white !important;
    margin-bottom: 18px;
    box-shadow: 0 8px 24px rgba(127,0,255,0.35);
}
#title-banner h1 { margin: 0; font-size: 2.1em; }
#title-banner p { margin: 6px 0 0 0; opacity: 0.95; }

.gradio-container { background: linear-gradient(160deg, #f7f3ff 0%, #fdf1fb 100%) !important; }

#draft-box textarea, #final-box textarea {
    border-radius: 14px !important;
    font-size: 15px !important;
    line-height: 1.55 !important;
}
#draft-box { border-left: 6px solid #A78BFA !important; border-radius: 14px; }
#final-box { border-left: 6px solid #EC4899 !important; border-radius: 14px; }

#input-box textarea {
    border-radius: 14px !important;
    border: 2px solid #E9D5FF !important;
}

footer {visibility: hidden}
"""

# ============================
# 🧠 Wraps the pipeline already built above (create_linkedin_post)
# ============================
def run_pipeline(user_information, progress=gr.Progress()):
    if not user_information or not user_information.strip():
        gr.Warning("من فضلك اكتب معلوماتك الأول 📝")
        return "", ""

    progress(0.2, desc="📝 الكاتب بيحضّر المسودة...")
    draft, final_post = create_linkedin_post(user_information)
    progress(1.0, desc="✅ خلصنا!")
    return draft, final_post


with gr.Blocks(theme=custom_theme, css=custom_css, title="AI LinkedIn Post Generator") as demo:

    gr.HTML(
        """
        <div id="title-banner">
            <h1>✨ AI LinkedIn Post Generator ✨</h1>
            <!-- <p>اكتب معلوماتك، وخلي الـ AI يكتبلك بوست احترافي جاهز للنشر 🚀</p> -->
        </div>
        """
    )

    with gr.Row():
        with gr.Column(scale=1):
            user_input = gr.Textbox(
                elem_id="input-box",
                label="📝 معلوماتك (Your Information)",
                placeholder="مثال\nI completed an AI project where I built...\nI used Python, LangChain and RAG.\nThe main thing I learned was...",
                lines=10,
            )
            generate_btn = gr.Button("🚀 Generate Post", variant="primary", size="lg")
            gr.Examples(
                examples=[
                    "I completed an AI project where I built a RAG chatbot.\nI used Python, LangChain and Hugging Face models.\nThe main thing I learned was how to evaluate retrieval quality.",
                    "I am learning Agentic AI.\nI am studying LLMs, RAG, AI agents, LangChain, FastAPI.\nI am working toward becoming a Full Stack AI Engineer.",
                ],
                inputs=user_input,
                label="💡 أمثلة سريعة",
            )

        with gr.Column(scale=1):
            draft_output = gr.Textbox(
                elem_id="draft-box",
                label="🟣 Draft (from Writer model)",
                lines=10
            )
            final_output = gr.Textbox(
                elem_id="final-box",
                label="🌸 Final Post (after Review)",
                lines=10
            )

    generate_btn.click(
        fn=run_pipeline,
        inputs=user_input,
        outputs=[draft_output, final_output],
    )

demo.launch(share=True, debug=True)
```
